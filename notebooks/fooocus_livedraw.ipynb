# 🎨 Fooocus-Lite: Interactive Image Generation

Welcome! This notebook sets up **Fooocus-Lite** in Google Colab for interactive AI image generation.

## What You'll Get
- ✅ A web interface for generating images from text prompts
- ✅ Optimized for Google Colab's T4 GPU (free tier)
- ✅ Beginner-friendly controls
- ✅ Real-time results (~30 seconds per image)

## Runtime
- **First run:** 5-10 minutes (downloads models)
- **Later runs:** 2-3 minutes (models cached)
- **Per image:** ~30 seconds

---

### ⚠️ Before You Start
1. Make sure you're using a **GPU runtime** (Runtime → Change runtime type → GPU)
2. All cells should run successfully; if any fail, restart runtime and try again
3. The Gradio interface link expires after ~1 hour (just re-run the launch cell)

## 🎨 You're All Set!

### How to Use
1. **Enter a prompt** in the text field (e.g., *"a dog wearing sunglasses"*)
2. **Adjust settings** if desired (resolution, guidance scale, etc.)
3. **Click Generate**
4. **Wait ~30 seconds** for your image
5. **Download** or use the interface to refine

### Tips
- ✅ Try simple prompts first (*"a cat"* before complex descriptions)
- ✅ Experiment with different styles: *"oil painting"*, *"photograph"*, *"3D render"*
- ✅ Use negative prompts to exclude unwanted elements
- ✅ Higher resolution = slower but better quality
- ✅ **Save outputs** by downloading from the interface or mounting Google Drive

### Troubleshooting
- **CUDA out of memory:** Runtime → Restart runtime. Reduce resolution in settings.
- **Link expired:** Re-run this cell to get a new one.
- **Slow generation:** First image is slower. Subsequent images use cached models.
- **Interface not loading:** Check browser console (F12) for errors. Try incognito mode.
